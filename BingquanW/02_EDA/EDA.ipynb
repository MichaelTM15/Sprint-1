```python

# Exploratory Data Analysis
# This section explores relationships between full-time match statistics
# and match outcomes in English football.
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
# ==============================================
# Exploratory Data Analysis
# Shots on Target Difference vs Match Result
# ==============================================

# 1. Load the half-cleaned dataset

from pathlib import Path

# 1. Load the half-cleaned dataset

from pathlib import Path
import pandas as pd

current = Path.cwd().resolve()

if (current / "processed-data").is_dir():
    ROOT = current
elif (current.parent / "processed-data").is_dir():
    ROOT = current.parent
elif (current.parent.parent / "processed-data").is_dir():
    ROOT = current.parent.parent
else:
    raise FileNotFoundError("Cannot find processed-data folder")

data_path = ROOT / "processed-data" / "football_data_halfclean.csv"

df = pd.read_csv(data_path, encoding="cp1252")
data = df

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print(df.head())

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print(df.head())
data = df

# 1. Load the half-cleaned dataset

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

current = Path.cwd().resolve()

if (current / "processed-data" / "football_data_halfclean.csv").is_file():
    ROOT = current
elif (current.parent / "processed-data" / "football_data_halfclean.csv").is_file():
    ROOT = current.parent
elif (current.parent.parent / "processed-data" / "football_data_halfclean.csv").is_file():
    ROOT = current.parent.parent
else:
    raise FileNotFoundError("Cannot find football_data_halfclean.csv")

data_path = ROOT / "processed-data" / "football_data_halfclean.csv"

df = pd.read_csv(data_path, encoding="cp1252")
data = df

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print(df.head())
# Folder for saving EDA figures
OUTPUT = ROOT / "BingquanW" / "02_EDA"
OUTPUT.mkdir(parents=True, exist_ok=True)
# 2. Select the variables needed for this analysis
plot_data = df[["HST", "AST", "FTR"]].copy()

# 3. Convert shots on target to numeric values
plot_data["HST"] = pd.to_numeric(
    plot_data["HST"], errors="coerce"
)

plot_data["AST"] = pd.to_numeric(
    plot_data["AST"], errors="coerce"
)

# 4. Remove rows with missing values
plot_data = plot_data.dropna(
    subset=["HST", "AST", "FTR"]
)

# Keep only valid match results
plot_data = plot_data[
    plot_data["FTR"].isin(["H", "D", "A"])
].copy()

# 5. Calculate shots-on-target difference
plot_data["ShotTargetDiff"] = (
    plot_data["HST"] - plot_data["AST"]
)

# 6. Create a boxplot
fig, ax = plt.subplots(figsize=(9, 6))

plot_data.boxplot(
    column="ShotTargetDiff",
    by="FTR",
    ax=ax,
    grid=False
)

ax.set_xlabel("Full-Time Result (A = Away, D = Draw, H = Home)")
ax.set_ylabel("Home Shots on Target - Away Shots on Target")
ax.set_title("Shots on Target Difference by Match Result")

plt.suptitle("")
plt.tight_layout()

# 7. Save the figure
plt.savefig("shots_on_target_boxplot.png", dpi=300)

plt.show()

# 8. Print summary statistics
print(
    plot_data.groupby("FTR")["ShotTargetDiff"].describe()
)


# ==========================================
# EDA Figure 2: Shot Difference vs Match Result
# ==========================================

# 1. Select the required columns
shot_data = df[["HS", "AS", "FTR"]].copy()

# 2. Convert shot statistics to numeric values
shot_data["HS"] = pd.to_numeric(
    shot_data["HS"], errors="coerce"
)

shot_data["AS"] = pd.to_numeric(
    shot_data["AS"], errors="coerce"
)

# 3. Remove missing values
shot_data = shot_data.dropna(
    subset=["HS", "AS", "FTR"]
)

# Keep only valid match results
shot_data = shot_data[
    shot_data["FTR"].isin(["H", "D", "A"])
].copy()

# 4. Calculate total shot difference
shot_data["ShotDiff"] = (
    shot_data["HS"] - shot_data["AS"]
)

# 5. Create the boxplot
fig, ax = plt.subplots(figsize=(9, 6))

shot_data.boxplot(
    column="ShotDiff",
    by="FTR",
    ax=ax,
    grid=False
)

ax.set_xlabel(
    "Full-Time Result (A = Away, D = Draw, H = Home)"
)

ax.set_ylabel(
    "Home Shots - Away Shots"
)

ax.set_title(
    "Shot Difference by Match Result"
)

plt.suptitle("")
plt.tight_layout()

# 6. Save the figure in 02_EDA

OUTPUT = ROOT / "BingquanW" / "02_EDA"
OUTPUT.mkdir(parents=True, exist_ok=True)

output_path = OUTPUT / "shot_difference_boxplot.png"

plt.savefig(OUTPUT / "shot_difference_boxplot.png", dpi=300)
plt.close()
plt.savefig(OUTPUT / "shot_difference_boxplot.png", dpi=300)
plt.close()

print("\nEDA Figure 2: Shot Difference")
print("Number of matches:", len(shot_data))

# 7. Print summary statistics
print(
    shot_data.groupby("FTR")["ShotDiff"].describe()
)

# ==========================================
# EDA Figure 3
# Corner Difference vs Match Result
# ==========================================

# Select relevant variables
corner_data = df[["HC", "AC", "FTR"]].copy()

# Convert corner statistics to numeric
corner_data["HC"] = pd.to_numeric(
    corner_data["HC"], errors="coerce"
)

corner_data["AC"] = pd.to_numeric(
    corner_data["AC"], errors="coerce"
)

# Remove missing values
corner_data = corner_data.dropna()

# Keep valid match results
corner_data = corner_data[
    corner_data["FTR"].isin(["H", "D", "A"])
].copy()

# Calculate corner difference
corner_data["CornerDiff"] = (
    corner_data["HC"] - corner_data["AC"]
)

# Create boxplot
fig, ax = plt.subplots(figsize=(9, 6))

corner_data.boxplot(
    column="CornerDiff",
    by="FTR",
    ax=ax,
    grid=False
)

ax.set_xlabel("Full-Time Result (A = Away, D = Draw, H = Home)")
ax.set_ylabel("Home Corners - Away Corners")
ax.set_title("Corner Difference by Match Result")

plt.suptitle("")
plt.tight_layout()

# Save figure
plt.savefig(
    OUTPUT / "corner_difference_boxplot.png",
    dpi=300
)

plt.show()

# Print summary statistics
print(
    corner_data.groupby("FTR")["CornerDiff"].describe()
)

# ==========================================
# EDA Figure 4: Home vs Away Shots on Target
# ==========================================

# 1. Select relevant columns
scatter_data = df[["FTR", "HST", "AST"]].copy()

# 2. Convert statistics to numeric values
scatter_data["HST"] = pd.to_numeric(
    scatter_data["HST"], errors="coerce"
)

scatter_data["AST"] = pd.to_numeric(
    scatter_data["AST"], errors="coerce"
)

# 3. Remove missing values and invalid results
scatter_data = scatter_data.dropna()
scatter_data = scatter_data[
    scatter_data["FTR"].isin(["H", "D", "A"])
]

# 4. Create scatter plot
plt.figure(figsize=(9, 6))

colors = {
    "H": "blue",
    "D": "green",
    "A": "red"
}

labels = {
    "H": "Home Win",
    "D": "Draw",
    "A": "Away Win"
}

for result in ["H", "D", "A"]:
    subset = scatter_data[scatter_data["FTR"] == result]

    plt.scatter(
        subset["HST"],
        subset["AST"],
        label=labels[result],
        color=colors[result],
        alpha=0.15,
        s=12
    )

# 5. Add labels
plt.xlabel("Home Shots on Target (HST)")
plt.ylabel("Away Shots on Target (AST)")
plt.title("Home vs Away Shots on Target by Match Result")

plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()

# 6. Save figure
plt.savefig(
    OUTPUT / "home_vs_away_shots_on_target.png",
    dpi=300
)
plt.show()

# 7. Print number of matches
print("EDA Figure 4: Home vs Away Shots on Target")
print("Number of matches:", len(scatter_data))
```
