# EDA and Modelling Plan

## Aim

My contribution focuses on exploratory data analysis (EDA) of football match statistics and developing Random Forest classification models.

The aim is to classify **Premier League match outcomes** using models trained on **Championship, League One and League Two** data. The target variable is **FTR**: H (home win), D (draw), or A (away win).

As the models use full-time statistics, this is **retrospective classification**, not pre-match forecasting.

---

## Dataset and EDA

The processed dataset contains **46,564 matches** across four English leagues (37,824 lower-league training matches and 8,740 Premier League test matches).

My EDA examines:
- Shots on target by match outcome
- Shot difference by match outcome
- Corner difference by match outcome
- Home versus away shots on target

The EDA helps explore which match statistics may be useful for classification.

---

## Data Preparation

I examine missing values using complete-case, available-case, mean imputation, regression imputation and multiple imputation approaches.

For the baseline Random Forest, complete-case analysis removes **3 training matches**. The odds-enhanced model removes **12 training matches**.

---

## Modelling Plan and Completed Experiments

- **Baseline Random Forest:** 12 full-time match statistics
- **Random Forest with Odds:** the same statistics plus B365H, B365D and B365A
- **Balanced Random Forest:** `class_weight="balanced"` to investigate poor draw classification

All models train on lower-league data and are evaluated on Premier League matches.

---

## Main Results and Next Steps

| Metric | Baseline RF | RF + Odds |
|---|---:|---:|
| Accuracy | 57.21% | 59.61% |
| Macro F1 | 45.76% | 47.41% |
| ROC-AUC (macro OvR) | 0.7134 | 0.7419 |
| Log Loss | 0.9240 | 0.8839 |

Class weighting improved baseline **draw recall from 6.21% to 20.52%**, while accuracy decreased to 55.76%.

The report discusses cross-league generalisation, poor draw classification, and the limitation of using full-time statistics. Future work could use a validation set for tuning.
