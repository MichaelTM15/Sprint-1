
# Random Forest Classification of Premier League Match Outcomes

## 1. Introduction

The research question for our group project is:

**How well can we classify outcomes from Premier League matches based on models trained on the lower leagues?**

This study investigates whether Random Forest models trained on lower English football leagues can successfully classify match outcomes in the Premier League.

Two Random Forest models are compared:

1. **Baseline Random Forest:** Uses 12 full-time match statistics.
2. **Random Forest with Odds:** Uses 12 full-time match statistics and three Bet365 betting odds features.

The models classify football results into three categories:

- **H:** Home win
- **D:** Draw
- **A:** Away win

The main aim is to examine how well models trained on Championship, League One and League Two data generalise to Premier League matches.

This project focuses on the retrospective classification of completed matches rather than forecasting results before kick-off, because full-time statistics are only available after a match has finished.
  


## 2. Dataset

### 2.1 Data Source

The dataset contains historical English football match data from Football-Data.co.uk.

The processed data was divided into two datasets based on football league divisions.

**Training Dataset:**
- Championship (E1)
- League One (E2)
- League Two (E3)

**Test Dataset:**
- Premier League (E0)

| Dataset | Number of Matches |
|---|---:|
| Lower-league training data | 37,824 |
| Premier League test data | 8,740 |

The models were trained using lower-league matches and evaluated using Premier League matches.

This design allows us to investigate whether relationships learned from lower-league football can generalise to the Premier League.

### 2.2 Input Features

The baseline Random Forest model uses 12 full-time match statistics:

| Feature | Description |
|---|---|
| HS | Home team shots |
| AS | Away team shots |
| HST | Home team shots on target |
| AST | Away team shots on target |
| HF | Home team fouls |
| AF | Away team fouls |
| HC | Home team corners |
| AC | Away team corners |
| HY | Home team yellow cards |
| AY | Away team yellow cards |
| HR | Home team red cards |
| AR | Away team red cards |

The second model also includes the following betting odds:

- B365H: Home win odds
- B365D: Draw odds
- B365A: Away win odds

The target variable is **FTR (Full-Time Result)**.

Full-time goals (FTHG and FTAG) are excluded from the predictors because they directly determine the match result.
  


## 3. Data Loading and Preparation

The processed dataset was prepared by a group member and divided according to football league divisions.

The training dataset contains matches from the Championship, League One and League Two, while the test dataset contains Premier League matches.

We load the two datasets separately to ensure that Premier League matches are not used to train the model.


```python

import pandas as pd
import numpy as np
from pathlib import Path

# Find the processed data folder
def find_data_folder():
    candidates = [
        Path.cwd() / "processed-data",
        Path.cwd().parent / "processed-data",
        Path.cwd().parent.parent / "processed-data"
    ]

    for folder in candidates:
        if (
            (folder / "football_train_data.csv").exists()
            and (folder / "football_test_data.csv").exists()
        ):
            return folder

    raise FileNotFoundError(
        "Cannot find the processed-data folder. "
        "Check the notebook's working directory."
    )

data_folder = find_data_folder()

# Load the datasets
train_data = pd.read_csv(
    data_folder / "football_train_data.csv"
)

test_data = pd.read_csv(
    data_folder / "football_test_data.csv"
)

print("Training dataset:", train_data.shape)
print("Premier League test dataset:", test_data.shape)

display(train_data.head())
```

```python

# Check target variable
print("Training outcome distribution:")
print(train_data["FTR"].value_counts())

print("\nPremier League outcome distribution:")
print(test_data["FTR"].value_counts())

# Compare outcome percentages
outcome_comparison = pd.DataFrame({
    "Lower leagues (%)": (
        train_data["FTR"]
        .value_counts(normalize=True) * 100
    ),
    "Premier League (%)": (
        test_data["FTR"]
        .value_counts(normalize=True) * 100
    )
}).round(2)

display(outcome_comparison)

# Check missing values
print("\nMissing values in training dataset:")
display(
    train_data.isna().sum().loc[
        lambda x: x > 0
    ]
)
```

## 5. Missing Data Analysis

### 5.1 Identifying Missing Values

Before training the Random Forest models, missing values were examined in the selected predictors.

The baseline model uses 12 full-time match statistics. Only three training observations contained missing values among these features.

Because the number of affected observations was very small compared with the full training dataset, several missing-data handling methods were investigated.

### 5.2 Missing Data Handling Methods

Five approaches were considered:

1. **Complete-case Analysis:** Removes observations containing missing values.
2. **Available-case Analysis:** Uses available observations for descriptive statistical calculations.
3. **Mean Imputation:** Replaces missing values with the mean of each feature.
4. **Regression Imputation:** Estimates missing values using relationships between variables.
5. **Multiple Imputation:** Generates multiple completed datasets to account for uncertainty in missing values.

Available-case analysis was used for descriptive analysis rather than as a separate Random Forest model.

### 5.3 Comparison of Methods

Four methods were compared using Random Forest classification.

| Method | Accuracy | Macro F1 | Draw Recall |
|---|---:|---:|---:|
| Complete-case | 57.21% | 45.76% | 6.21% |
| Mean Imputation | 56.97% | 45.57% | 6.25% |
| Regression Imputation | 56.99% | 45.54% | 6.07% |
| Multiple Imputation | 57.16% | 45.61% | 5.93% |

Complete-case analysis achieved the highest overall accuracy and Macro F1-score, although the differences between methods were small.

Therefore, complete-case analysis was selected for the baseline Random Forest model.

This choice was appropriate for the current dataset because only three training observations were affected. However, it does not mean that complete-case analysis is always better than imputation.


```python

# Selected features for the baseline Random Forest
features = [
    "HS", "AS", "HST", "AST",
    "HF", "AF", "HC", "AC",
    "HY", "AY", "HR", "AR"
]

# Missing values in the selected features
missing_summary = pd.DataFrame({
    "Training Missing": train_data[features].isna().sum(),
    "Test Missing": test_data[features].isna().sum()
})

display(missing_summary)

# Number of observations containing missing features
train_missing_rows = train_data[features].isna().any(axis=1).sum()
test_missing_rows = test_data[features].isna().any(axis=1).sum()

print("Training rows with missing features:", train_missing_rows)
print("Test rows with missing features:", test_missing_rows)

# Complete-case dataset
train_clean = train_data.dropna(
    subset=features + ["FTR"]
).copy()

test_clean = test_data.dropna(
    subset=features + ["FTR"]
).copy()

print("Clean training dataset:", train_clean.shape)
print("Clean test dataset:", test_clean.shape)
```


## 6. Random Forest Model

### 6.1 Model Implementation

Random Forest is an ensemble learning algorithm that combines multiple decision trees to improve classification performance.

The baseline Random Forest model uses 12 full-time match statistics to classify match outcomes into home wins, draws and away wins.

The model is trained on Championship, League One and League Two matches and evaluated on Premier League matches.

The following hyperparameters are used:

| Hyperparameter | Value |
|---|---|
| Number of trees (`n_estimators`) | 200 |
| Maximum depth (`max_depth`) | 12 |
| Random state | 42 |
| Parallel processing (`n_jobs`) | -1 |

These parameters are kept consistent with the original Random Forest experiment.


```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score

# Prepare training and testing features
X_train_clean = train_clean[features]
y_train_clean = train_clean["FTR"]

X_test = test_clean[features]
y_test = test_clean["FTR"]

# Create Random Forest model
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

# Train on lower-league matches
model.fit(X_train_clean, y_train_clean)

# Predict Premier League match outcomes
y_pred = model.predict(X_test)

# Evaluate model
accuracy = accuracy_score(y_test, y_pred)
macro_f1 = f1_score(y_test, y_pred, average="macro")

print(f"Premier League Accuracy: {accuracy:.4f}")
print(f"Premier League Macro F1: {macro_f1:.4f}")
```


### 6.2 Training and Test Performance

To investigate how well the Random Forest model generalises across football leagues, its classification performance is compared between the lower-league training dataset and the Premier League test dataset.

The training accuracy shows how well the model classifies the data used for fitting, while the test accuracy measures performance on Premier League matches that were not used during training.

A difference between training and test accuracy may reflect overfitting, differences in match characteristics, or differences in the distribution of match outcomes across leagues.

Both accuracy and Macro F1-score are reported because the model has difficulty identifying drawn matches.


```python

from sklearn.metrics import accuracy_score, f1_score
import pandas as pd

# Predictions on lower-league training data
train_pred = model.predict(X_train_clean)

# Predictions on Premier League test data
test_pred = model.predict(X_test)

# Compare performance
performance_comparison = pd.DataFrame({
    "Dataset": [
        "Lower-league Training",
        "Premier League Testing"
    ],
    "Accuracy": [
        accuracy_score(y_train_clean, train_pred),
        accuracy_score(y_test, test_pred)
    ],
    "Macro F1": [
        f1_score(y_train_clean, train_pred, average="macro"),
        f1_score(y_test, test_pred, average="macro")
    ]
})

display(performance_comparison.round(4))

accuracy_gap = (
    performance_comparison.loc[0, "Accuracy"]
    - performance_comparison.loc[1, "Accuracy"]
)

print(f"Training - Test Accuracy Gap: {accuracy_gap:.4f}")
```


### 6.3 Confusion Matrix

A confusion matrix is used to examine how well the Random Forest model identifies home wins, draws and away wins.

It provides more information than overall accuracy because it shows which outcome categories are frequently misclassified.

This is particularly important for drawn matches, which the model has difficulty identifying.


```python

import matplotlib.pyplot as plt
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    confusion_matrix,
    classification_report
)

labels = ["H", "D", "A"]

cm = confusion_matrix(
    y_test,
    test_pred,
    labels=labels
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Home Win", "Draw", "Away Win"]
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Random Forest: Premier League Confusion Matrix")
plt.tight_layout()
plt.show()

print("Classification Report:")
print(
    classification_report(
        y_test,
        test_pred,
        labels=labels,
        target_names=["Home Win", "Draw", "Away Win"],
        zero_division=0
    )
)
```


### 6.4 Handling Class Imbalance

The baseline Random Forest model has difficulty identifying drawn matches, with a draw recall of only 6.21%.

To investigate this problem, we examine the distribution of home wins, draws and away wins in the training dataset.

We then compare the original Random Forest with a class-weighted Random Forest using `class_weight="balanced"`.

This method gives greater importance to less frequent classes during model training without changing the Premier League test dataset.


```python

# Check class distribution
class_distribution = pd.DataFrame({
    "Count": y_train_clean.value_counts(),
    "Percentage (%)": (
        y_train_clean.value_counts(normalize=True) * 100
    ).round(2)
})

display(class_distribution)
```

```python


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    recall_score,
    precision_score
)
import pandas as pd

# Random Forest with balanced class weights
rf_balanced = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Train on lower-league data only
rf_balanced.fit(X_train_clean, y_train_clean)

# Predict Premier League outcomes
y_pred_balanced = rf_balanced.predict(X_test)

# Compare the original and balanced models
def evaluate_model(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro F1": f1_score(
            y_true, y_pred, average="macro"
        ),
        "Draw Precision": precision_score(
            y_true == "D", y_pred == "D",
            zero_division=0
        ),
        "Draw Recall": recall_score(
            y_true == "D", y_pred == "D",
            zero_division=0
        ),
        "Draw F1": f1_score(
            y_true == "D", y_pred == "D",
            zero_division=0
        )
    }

comparison = pd.DataFrame({
    "Original RF": evaluate_model(y_test, y_pred),
    "Balanced RF": evaluate_model(y_test, y_pred_balanced)
}).T

display((comparison * 100).round(2))
```

```python

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

fig, ax = plt.subplots(figsize=(7, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_balanced,
    labels=["H", "D", "A"],
    display_labels=["Home Win", "Draw", "Away Win"],
    cmap="Blues",
    values_format="d",
    ax=ax
)

ax.set_title("Balanced Random Forest - Premier League")
plt.tight_layout()
plt.show()
```


### 6.5 Evaluation of Class Weight Balancing

To address the low recall for drawn matches, a second Random Forest was trained using `class_weight="balanced"`.

This approach adjusts the importance of each class according to its frequency in the training data.

The balanced model was evaluated against the original Random Forest using Accuracy, Macro F1, Draw Precision, Draw Recall and Draw F1.

The main objective was to investigate whether class weighting could improve the identification of drawn matches and how this affected overall classification performance.

The results are presented above. Improvements in draw recall must be considered alongside changes in precision, Macro F1 and overall accuracy.


```python

import numpy as np
import matplotlib.pyplot as plt

# Use the comparison table calculated earlier
metrics = [
    "Accuracy",
    "Macro F1",
    "Draw Precision",
    "Draw Recall",
    "Draw F1"
]

original = comparison.loc["Original RF", metrics].to_numpy() * 100
balanced = comparison.loc["Balanced RF", metrics].to_numpy() * 100

x = np.arange(len(metrics))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 6))

bars1 = ax.bar(
    x - width / 2,
    original,
    width,
    label="Original RF"
)

bars2 = ax.bar(
    x + width / 2,
    balanced,
    width,
    label="Balanced RF"
)

ax.set_ylabel("Score (%)")
ax.set_title(
    "Original vs Balanced Random Forest - Premier League"
)
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.set_ylim(0, 75)
ax.legend()

ax.bar_label(bars1, fmt="%.2f", padding=3)
ax.bar_label(bars2, fmt="%.2f", padding=3)

plt.tight_layout()
plt.show()
```


### Comparison of Original and Balanced Random Forest

The figure compares classification performance before and after applying class weighting.

Draw recall increased from **6.21% to 20.52%**, while draw F1-score improved from **10.75% to 25.24%**.

Macro F1-score also increased from **45.76% to 50.24%**, suggesting improved performance across the three outcome categories.

However, overall accuracy decreased from **57.21% to 55.76%**, and draw precision decreased from **40.12% to 32.79%**.

This demonstrates a precision–recall trade-off: the balanced model identifies more actual draws but also incorrectly predicts more draws.

Although class weighting improved draw classification, its recall remained relatively low at 20.52%, indicating that further improvements are needed.




## 7. Random Forest with Betting Odds

### 7.1 Model Description

The second Random Forest model includes three additional Bet365 betting odds features alongside the original 12 full-time match statistics.

The additional features are:

- **B365H:** Bet365 home win odds
- **B365D:** Bet365 draw odds
- **B365A:** Bet365 away win odds

The purpose of adding betting odds is to investigate whether information from the betting market improves the classification of Premier League match outcomes.

The same Random Forest hyperparameters are used to allow a direct comparison with the baseline model:

- `n_estimators = 200`
- `max_depth = 12`
- `random_state = 42`
- `n_jobs = -1`

The model is trained on Championship, League One and League Two matches and evaluated on Premier League matches.

Since this model also includes full-time match statistics, it remains a retrospective classification model rather than a genuine pre-match forecasting model.


```python

# Original 12 match statistics + 3 betting odds
odds_features = features + ["B365H", "B365D", "B365A"]

# Remove rows missing any required predictor or target
train_odds_clean = train_data.dropna(
    subset=odds_features + ["FTR"]
).copy()

test_odds_clean = test_data.dropna(
    subset=odds_features + ["FTR"]
).copy()

# Prepare training and test datasets
X_train_odds = train_odds_clean[odds_features]
y_train_odds = train_odds_clean["FTR"]

X_test_odds = test_odds_clean[odds_features]
y_test_odds = test_odds_clean["FTR"]

print("Number of features:", len(odds_features))
print("Odds model training rows:", len(X_train_odds))
print("Odds model test rows:", len(X_test_odds))
print("Removed training rows:", len(train_data) - len(train_odds_clean))
print("Removed test rows:", len(test_data) - len(test_odds_clean))
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score

# Create Random Forest with betting odds
rf_odds = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

# Train on lower-league matches
rf_odds.fit(X_train_odds, y_train_odds)

# Predict Premier League outcomes
y_pred_odds = rf_odds.predict(X_test_odds)

# Evaluate the model
odds_accuracy = accuracy_score(y_test_odds, y_pred_odds)
odds_macro_f1 = f1_score(
    y_test_odds,
    y_pred_odds,
    average="macro"
)

print(f"RF with Odds Accuracy: {odds_accuracy:.4f}")
print(f"RF with Odds Macro F1: {odds_macro_f1:.4f}")
```

```python

model_comparison = pd.DataFrame({
    "Model": [
        "Random Forest (12 features)",
        "Random Forest with Odds (15 features)"
    ],
    "Accuracy (%)": [
        accuracy * 100,
        odds_accuracy * 100
    ],
    "Macro F1 (%)": [
        macro_f1 * 100,
        odds_macro_f1 * 100
    ]
})

display(model_comparison.round(2))
```


### 7.2 Probability-Based Model Evaluation

In addition to accuracy and Macro F1-score, the models are evaluated using multiclass ROC-AUC and Log Loss.

Multiclass ROC-AUC measures how well the models distinguish between home wins, draws and away wins based on predicted probabilities.

Log Loss evaluates the quality of the predicted probabilities. A lower Log Loss indicates better probabilistic performance.

These metrics provide additional information beyond classification accuracy.


```python

from sklearn.metrics import roc_auc_score, log_loss

# Predict probabilities for both models
baseline_proba = model.predict_proba(X_test)
odds_proba = rf_odds.predict_proba(X_test_odds)

# Multiclass ROC-AUC
baseline_auc = roc_auc_score(
    y_test,
    baseline_proba,
    labels=model.classes_,
    multi_class="ovr",
    average="macro"
)

odds_auc = roc_auc_score(
    y_test_odds,
    odds_proba,
    labels=rf_odds.classes_,
    multi_class="ovr",
    average="macro"
)

# Multiclass Log Loss
baseline_logloss = log_loss(
    y_test,
    baseline_proba,
    labels=model.classes_
)

odds_logloss = log_loss(
    y_test_odds,
    odds_proba,
    labels=rf_odds.classes_
)

# Comparison table
probability_comparison = pd.DataFrame({
    "Model": [
        "Random Forest",
        "Random Forest + Odds"
    ],
    "Multiclass ROC-AUC": [
        baseline_auc,
        odds_auc
    ],
    "Multiclass Log Loss": [
        baseline_logloss,
        odds_logloss
    ]
})

display(probability_comparison.round(4))
```


## 8. Feature Importance

Feature importance analysis was conducted to investigate which variables contributed most to the Random Forest classifications.

The baseline model contains 12 full-time match statistics, while the odds-enhanced model contains 15 predictors, including three Bet365 betting odds features.

Random Forest provides impurity-based feature importance scores, which describe how much each feature contributes to reducing impurity across the decision trees.

However, these scores do not establish causal relationships and may be affected by correlated predictors.


```python

import pandas as pd
import matplotlib.pyplot as plt

# Baseline Random Forest feature importance
importance_baseline = pd.Series(
    model.feature_importances_,
    index=features
).sort_values(ascending=True)

# Random Forest with Odds feature importance
importance_odds = pd.Series(
    rf_odds.feature_importances_,
    index=odds_features
).sort_values(ascending=True)

# Baseline plot
fig, ax = plt.subplots(figsize=(9, 6))
importance_baseline.plot.barh(ax=ax)

ax.set_title("Feature Importance - Baseline Random Forest")
ax.set_xlabel("Feature Importance")
ax.set_ylabel("Feature")

plt.tight_layout()
plt.show()

# Odds model plot
fig, ax = plt.subplots(figsize=(9, 7))
importance_odds.plot.barh(ax=ax)

ax.set_title("Feature Importance - Random Forest with Odds")
ax.set_xlabel("Feature Importance")
ax.set_ylabel("Feature")

plt.tight_layout()
plt.show()

# Display top 5 predictors for each model
print("Top 5 features - Baseline RF:")
display(
    importance_baseline.sort_values(
        ascending=False
    ).head(5).round(4)
)

print("Top 5 features - RF with Odds:")
display(
    importance_odds.sort_values(
        ascending=False
    ).head(5).round(4)
)
```


### 8.1 Interpretation of Feature Importance

The feature importance plots compare the relative contributions of match statistics and betting odds in the two Random Forest models.

The baseline model uses only full-time match statistics, while the second model also includes Bet365 opening odds.

By comparing the rankings, we can investigate whether betting odds become important predictors and whether the relative importance of match statistics changes after adding these features.

Feature importance describes associations learned by the models rather than causal effects.



### 8.2 Feature Importance Results

In the baseline Random Forest model, the most important features were home shots on target (HST, 22.45%) and away shots on target (AST, 19.97%).

Other important predictors included home corners (HC, 8.10%), home shots (HS, 8.03%) and away shots (AS, 7.68%).

In the Random Forest model with betting odds, HST (17.10%) and AST (15.49%) remained the two most important features.

However, Bet365 home win odds (B365H, 9.32%) and away win odds (B365A, 9.23%) became the third and fourth most important predictors.

These results suggest that betting odds provide additional information alongside full-time match statistics.

The relative importance of shots on target decreased after adding odds, partly because importance was redistributed across a larger set of predictors.

Overall, shots on target were consistently important in both models, while betting odds also contributed meaningfully to the odds-enhanced model.

Feature importance reflects patterns learned by Random Forest rather than causal relationships.



## 9. Discussion

### 9.1 Comparison of Random Forest Models

This study compared two Random Forest models trained on Championship, League One and League Two matches and evaluated on Premier League matches.

The baseline Random Forest used 12 full-time match statistics and achieved an accuracy of 57.21%, a Macro F1-score of 45.76%, and a multiclass ROC-AUC of 0.7134.

After including three Bet365 betting odds features, accuracy increased to 59.61%, Macro F1 increased to 47.41%, and multiclass ROC-AUC increased to 0.7419.

Multiclass Log Loss also decreased from 0.9240 to 0.8839.

These results suggest that betting odds provide additional information that improves overall match-outcome classification.

### 9.2 Cross-League Generalisation

The baseline Random Forest achieved a training accuracy of 66.20%, compared with 57.21% on the Premier League test dataset.

This difference of 8.99 percentage points suggests that the model does not generalise perfectly from lower-league football to the Premier League.

The gap may reflect overfitting, differences in match characteristics, or changes in the distribution of outcomes across leagues.

For example, draws represented 27.02% of the lower-league dataset but only 24.70% of Premier League matches. This difference could influence overall accuracy because the model struggles to identify draws.

### 9.3 Class Imbalance and Draw Classification

A major weakness of the baseline model was its low draw recall of 6.21%.

To investigate this issue, a class-weighted Random Forest was trained using `class_weight="balanced"`.

Draw recall improved from 6.21% to 20.52%, and Draw F1 increased from 10.75% to 25.24%.

Macro F1 also improved from 45.76% to 50.24%.

However, overall accuracy decreased from 57.21% to 55.76%, while Draw Precision declined from 40.12% to 32.79%.

These results demonstrate that improving the identification of drawn matches can involve a trade-off between precision and recall.

The class-weighted model provided better balanced performance across the three outcomes, although most draws were still misclassified.

### 9.4 Feature Importance

Shots on target were the most important predictors in both Random Forest models.

In the baseline model, HST and AST contributed 22.45% and 19.97% of total feature importance, respectively.

After adding odds, B365H and B365A became the third and fourth most important features.

These findings suggest that both attacking statistics and betting market information contributed to the models' classifications.



## 10. Limitations

### 10.1 Poor Draw Classification

Both Random Forest models struggled to identify drawn matches.

Although class weighting improved draw recall in the baseline model, its recall remained relatively low at 20.52%.

Further research could investigate alternative class weights, resampling methods and probability thresholds.

### 10.2 Use of Full-Time Match Statistics

The models use full-time statistics such as shots, corners and fouls.

These features are only completely available after a match has finished.

Therefore, the models should be interpreted as retrospective match-outcome classifiers rather than genuine pre-match forecasting models.

A model intended for pre-match prediction would need to use information available before kick-off, such as opening odds and historical team performance.

### 10.3 Cross-League Differences

The models were trained on lower-league matches and evaluated on Premier League matches.

Differences in playing styles, team quality and outcome distributions may affect how well patterns transfer between leagues.

This could partly explain the difference between training and test performance.

### 10.4 Model Selection and Validation

The Random Forest models used fixed hyperparameters rather than an extensive hyperparameter search.

The class-weighted experiment was also compared using the Premier League test dataset.

Therefore, the results of the balancing experiment should be interpreted as exploratory rather than evidence of a fully validated model selection procedure.

Future work should select hyperparameters and class weights using a validation set drawn from the lower-league training data, keeping the Premier League dataset separate for final evaluation.

### 10.5 Missing Data

Complete-case analysis was used because relatively few observations had missing values among the selected predictors.

However, removing observations may introduce bias if the missing data are not random.

Alternative imputation approaches produced similar results in the baseline Random Forest experiment.


## 11. Conclusion

This project investigated whether Random Forest models trained on lower English football leagues could classify match outcomes in the Premier League.

The baseline model used 12 full-time match statistics, while the second model included three additional Bet365 betting odds features.

The baseline Random Forest achieved an accuracy of **57.21%**, a Macro F1-score of **45.76%**, and a multiclass ROC-AUC of **0.7134**.

Adding betting odds improved accuracy to **59.61%**, Macro F1 to **47.41%**, and multiclass ROC-AUC to **0.7419**. Log Loss also decreased from **0.9240 to 0.8839**.

These findings suggest that betting odds provide useful additional information for retrospective match classification.

A major limitation was the difficulty of classifying drawn matches. Applying balanced class weights increased draw recall from **6.21% to 20.52%** and Draw F1 from **10.75% to 25.24%**, although overall accuracy decreased slightly.

The results also showed an **8.99 percentage-point gap** between baseline training accuracy and Premier League test accuracy, indicating challenges in cross-league generalisation.

Overall, Random Forest demonstrated some ability to transfer patterns learned from lower-league football to Premier League match classification, with the odds-enhanced model achieving the best overall performance among the two main models.

Future work could focus on class imbalance, hyperparameter optimisation, cross-validation and pre-match feature engineering.
