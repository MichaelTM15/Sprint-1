## The $d$-invariant

### Question 1

Before writing the function to calculate $d(S(p,q),i)$, there are a few things to deal with.
The first is that python will not return an exact fraction for a recurring decimal; intead it will output it as a decimal rounded to 17 d.p. The standard Python library has a function for this:

```python
from fractions import Fraction

def decimal_to_fraction(decimal_number): 
    return Fraction(decimal_number).limit_denominator() #Fraction().limit_denominator finds rational approximation for given floating point number
## returns Fraction(a,b) = a/b
```

We also need to check if $p,q$ are coprime. This is equivalent to having $gcd(p,q)=1$, so the `gcd` function is useful: 

```python
def gcd(a, b):
    # Inputs: integers a, b"
    # Returns the greatest common divisor of a, b"
    if not(a % 1 == 0  and b % 1 == 0):
        print("Only integer inputs permitted.")
        return None
    if b == 0: 
        return abs(a)
    else:
        return gcd(b, a % b)
```

Now the $d$-invariant function can be written. It will be recursive, so we will use memoization for it.

```python
mem=dict()
## dictionary for memoization

def d(p, q, i):
    global mem
    if p==1:
        mem[p,q,i] = 0 # a base case
    if gcd(p,q) != 1:
        return None # not a valid input
    else:
        if (p, q, i) not in mem:
        ## main part of the function
            if p>q>0 and 0<=i<(p+q):
                mem[p,q,i] = ((2*i+1 - p - q)**2 - p*q)/(4*p*q) - d(q,p,i) # when conditions are satisfied it runs recursively
            else:
                mem[p,q,i] = d(p,q%p,i%p) # exchanging q and i with q%p and i%p (which is equivalent but now satisfies the above conditions) 
        return decimal_to_fraction(mem[p,q,i]) # returns an exact fraction instead of rounding
```

To check this runs as intended, we can compare its output for $S(p,q)=S(9,7)$ to the given values.

```python
print(d(9,7,0),d(9,7,1),d(9,7,2),d(9,7,3), d(9,7,4), d(9,7,5),d(9,7,6),d(9,7,7),d(9,7,8))
```

Hence the function is working as it should.

### Question 5

Let us call the set of numbers $\frac{m^2}{q}$ that have this property __M__. Firstly we want the list of odd numbers $m<105$ to test for a given $q$.

```python
m = []
for i in range(0,105):
    if i%2 == 1:
        m.append(i)
    else:
        None
```

The following function outputs the set of fractions $\frac{m^2}{q}$ such that the knot $S(m^2,q)$ has $m$ values of the $d$-invariants that are equal to 0. The value for $q$ is the input - this is done as a way of splitting up the set of numbers that satisfy the given conditions.

```python
def d_zeros(q):
    zeros = [] # begins as empty list
    for i in range(0, len(m)):
        inv = []
        for j in range(0, m[i]**2): # the length of the d-invariant list for S(m^2,q), where m = m[i] each loop
            if d((m[i])**2, q,j) == 0: 
                inv.append(1)
                ## this records each time that one of the d-invariants are zero
        if len(inv) == m[i]:
            zeros.append(decimal_to_fraction(m[i]**2/q))
            ## if the number of zeros == m, then we want to record m^2/q
    return zeros
```

It can be seen that, for $q=4$, there is 1 $d$-invariant equal to 0 for $m^2=1$, 3 for $m^2=9$ and 5 for $m^2=25$. This indicates that the outputs are valid; but cannot confirm that it every possible fraction is given.

In order to compare our list(s) to it, we should define a function that outputs the set __R__ in [1, Definition 1.1] for a given $q$ and with $p=m^2$ for odd $m<105$.

The set is defined as follows:

"*__R__ contains the set of rational numbers $\frac{p}{q}$ such that $p > q > 0$, $gcd(p, q) = 1$, $p = m^2$ for some $m ∈ N$ and $q$ is of one of the following types:*

*(1) $mk ± 1$ with $m > k > 0$ and $gcd(m, k) = 1$;*

*(2) $d(m ± 1)$, where $d > 1$ divides $2m ∓ 1$;*

*(3) $d(m ± 1)$, where $d > 1$ is odd and divides $m ± 1$.*"

```python
def R_list(q): # again we split up the set based on what q is - this can be used for direct comparison with M
    R = []
    for i in range(0,len(m)): # going through each possible value of p
        if gcd(m[i]**2,q) ==1 and (m[i]**2)>q>0:
        ## initial conditions, then sorted by if q is any of the three types described above
            for k in range(1,m[i]): # (1)
                if k*m[i] +1 ==q or k*m[i] - 1 == q:
                    if gcd(m[i],k) ==1:
                        R.append(decimal_to_fraction(m[i]**2/q))
            
            for j in range(2,208): # (2)
                ## 207 is the greatest value that 2m+1 can be, as max(m)=103
                if j*(m[i]+1)==q and (2*m[i]-1)%j == 0:
                    R.append(decimal_to_fraction(m[i]**2/q))
                if j*(m[i]-1)==q and (2*m[i]+1)%j == 0:
                    R.append(decimal_to_fraction(m[i]**2/q))
            
            for l in range(2,208): # (3)
                if l%2 == 1:
                    if (m[i]+1)%l == 0 and l*(m[i]+1)==q:
                        R.append(decimal_to_fraction(m[i]**2/q))
                    if (m[i]-1)%l == 0 and l*(m[i]-1)==q:
                        R.append(decimal_to_fraction(m[i]**2/q))
    return R
```

Now we can compare the two sets in a few examples:

```python
print("R contains " +str(R_list(2)))
print("M contains " +str(d_zeros(2)))
print()
print("R contains " +str(R_list(100)))
print("M contains " +str(d_zeros(100)))
```

It is clear that as $d(S(1,q),i)=0$ for any $q,i$, the fraction $\frac{1}{q}$ will be part of the set __M__. However, __R__ will not include any fraction with numerator 1; as $p$ must be greater than $q$ for its conditions to hold. Furthermore this means that any other potential fractions that are in __M__ might not be in __R__ due to $p$ being smaller than $q$.

However these are not the only numbers missing from __R__. Consider when $q=37$:

```python
print(R_list(37))
print(d_zeros(37))
```

$\frac{121}{37}$ is not in __R__ even though it is greater than 1. 

Therefore we can at most say that __R__ is a subset of __M__. To test this we will randomly generate a value of $q$ several times.

```python
for i in range(0,5): # 5 tests
    q = randrange(1,100) # randomly pick q and get the lists
    r=R_list(q)
    z=d_zeros(q)
    for k in range(0,len(r)):
        if r[k] not in z:
            print("ERROR")
            ## all elements in R should be in M - no ERROR messages indicates this
```

Thus we conclude that all elements in __R__ satisfy the same properties as those in __M__, hence __R__$\subseteq$__M__.
